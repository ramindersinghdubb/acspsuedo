### Viewing Dataset/Variable Metadata: American Community Survey 1-Year Public Use Microdata Sample

For end-users, you may be unsure as to what type of data you might be looking for.

<br>

To that end, `'acspsuedo.query'` provides a friendly `variable_cache` interface to view metadata on all available variables and tables for a dataset.

*Note: Navigating dataset metadata is purposefully conducted via this interface since this interface is employed as part of making queries to the Census Bureau. By pre-exploring the data, users can more quickly and correctly send queries to the Census Bureau.*

```python
import acspsuedo.query as apq

variable_cache = apq.variable_cache

from acspsuedo.datasets import ACS1_PUMS

DATASET = ACS1_PUMS
YEAR = 2014
```

#### Fetching metadata on all variables

The `var_metadata_df()` method of the interface provides holistic information on all variables supported by the dataset in question, with such information covering:

- The name of the variable

- The variable's corresponding label

- The data type of the variable, and

- What tables contain that variable

```python
VARIABLE_METADATA_DF = variable_cache.var_metadata_df(DATASET, YEAR)

VARIABLE_METADATA_DF
```

*Note: Public use microdata samples do not have topics or tables. However, for other datasets, such as the American Community Survey 1-Year or 5-Year Detailed Tables, this is the case.*

Suppose you might be interested in variables covering earnings or income. You can find variables pertaining to those topics as so.

```python
TOPICS = ['income', 'earnings']

VARIABLE_METADATA_DF[
    VARIABLE_METADATA_DF['LABEL'].str.contains('|'.join(TOPICS))
]
```

From there, you can run queries for data or continue to explore the metadata further.

In our case, we will choose the `WAGP` (use ADJINC to adjust WAGP to constant dollars) and the `PAP` (PUMS SSI/AFDC/other welfare income) variables to run a mock query.

Also, since the PUMS data covers record-like information on individual persons and housing units, it is recommended to add two identification variables:
- `SERIALNO` (Housing unit/GQ person serial number)
- `SPORDER` (Person number)

*For further reference on PUMS guidelines, see [https://www.census.gov/programs-surveys/acs/microdata/access.html](https://www.census.gov/programs-surveys/acs/microdata/access.html)*

```python
SELECTED_VARIABLES = [
    'SERIALNO',
    'SPORDER',
    'WAGP',
    'PAP'
]
```

```python
# First, view supported geographic specifiers for our dataset

apq.view_geographic_paths(DATASET, YEAR)
```

```python
# We select ['state', 'public_use_microdata_area'] and run our query

from acspsuedo.fips.states import CA # <- Abbreviation for California

df = apq.download(
    dataset = DATASET,
    year = YEAR,
    variables = SELECTED_VARIABLES,
    state = CA,
    public_use_microdata_area = '*'
)

df
```

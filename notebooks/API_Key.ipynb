### Setting up an API Key

Often, especially for dashboard-based applications and/or research investigations, users may be interested in recurrent/large queries to the Census Bureau. However, running multiple (500+) queries to the Census Bureau *without* an API key will limit a user's ability to query in a single session.

Thus, the Census Bureau recommends the use of an API key. Users and organizations can request an API key for free at [https://api.census.gov/data/key_signup.html](https://api.census.gov/data/key_signup.html).

`acspsuedo` allows for users to supply their API key in three different ways through the `acspsuedo.query.api_key_config` interface.

1. Setting the API key directly via the `api_key_config.API_KEY` priority (**Prioritized**)

2. Setting the API key via the operating system environment

3. Setting the API key in a textfile written in the current working directory

```python
import acspsuedo.query as apq

api_key_config = apq.api_key_config
```

#### 1. Set an API key directly
The `api_key_config` interface takes an `API_KEY` property that you can directly set.

```python
api_key_config.API_KEY = 'Your_API_key'
```

#### 2. Set the API key in the operating system environment

If an API key is not manually set, the `api_key_config` interface will first try looking in the operating system (OS) environment.

The default OS location is `'CENSUS_BUREAU_API_KEY'`. The location to the API key can be customized by setting the `OS_ENV_LOCATION` property of the interface.

```python
import os

# Default OS location
os.environ['CENSUS_BUREAU_API_KEY'] = 'Your_API_key'


env_location = api_key_config.OS_ENV_LOCATION
print('Environment location:', api_key_config.OS_ENV_LOCATION)

supplied_api_key = os.environ.get(env_location, '')
print('Supplied API key:', supplied_api_key)
```

```python
# Customize the OS location
api_key_config.OS_ENV_LOCATION = 'NEW_OS_LOCATION_TO_API_KEY'

# Setting the API key accordingly
new_env_location = api_key_config.OS_ENV_LOCATION
print('New environment location:', new_env_location)

os.environ['NEW_OS_LOCATION_TO_API_KEY'] = 'Your_API_key'

api_key = os.environ.get(new_env_location, '')
print('Supplied API key:', supplied_api_key)
```

#### 3. Set the API key in a textfile of the current working directory.

If all else does not suffice, the `api_key_config` interface will default to looking for a textfile called `api_key.txt` in the current working directory (`./api_key.txt`).

If this textfile does not exist, the `API_KEY` property will be left as an empty string value.

<br>

The location of this textfile can be customized by setting the `FILE_PATH` property of the interface.

```python
# Default textfile location
print(api_key_config.FILE_PATH)
```

```python
# New textfile location
api_key_config.FILE_PATH = '/folder/to/some/new_file_path_location.txt'

print(api_key_config.FILE_PATH)
```

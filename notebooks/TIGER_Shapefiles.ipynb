### TIGER Shapefiles

For end-users, we have implemented a TIGER shapefile handler interface that performs two basic purposes.

1. Fetching TIGER shapefiles from the Census Bureau database.

2. Locally caching TIGER shapefiles on the user's operating system (*provided caching is the behavior; by default, this is the case*).

While users are generally discouraged from not changing the core functionalities of (1), you can customize whether or not caching is enabled, the location of the cache, and (if the cache location is updated) whether or not pre-existing shapefiles should be tracked/moved into the new cache.

```python
import acspsuedo.query as apq

print("Default caching behavior:", apq.shapefile_handler.auto_cache)
```

The interface is `shapefile_handler`, whose caching behaviors can be customized via its attributes:

- `shapefile_handler.auto_cache` (boolean; default `True`)

    Indicate whether or not to automatically cache extracted shapefile information.

    <br>

- `shapefile_handler.cache_path` (`pathlib.Path` or string; default `Path.home() / 'cache' / 'acspsuedo' / 'TIGER_shapefiles'`)
    
    If `auto_cache` is True, `cache_path` specifies the caching location of extracted shapefiles.

    <br>

- `shapefile_handler.track_updated_cache` (boolean; default `True`)
    
    Indicate whether or not tracked shapefiles posited in the previous cache location should be moved if/when a new cache location should be specified.

    With this implementation, it is assumed that the previous cached folder may be relevant beyond the immediate use within this overall architecture (i.e. to cache TIGER shapefiles that can be reached with this interface). For instance, users may want to cache TIGER shapefiles that they themselves have manually downloaded or were not otherwise accessible from the functionalities of this handler.

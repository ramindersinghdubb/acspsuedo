### Viewing Geographic Specification Metadata: American Community Survey, 1-Year Estimate Detailed Tables

For end-users, you may be unsure as to what geographic specifications to supply for a desired set of data.

<br>

To that end, `acspsuedo.query` provides two helper functions to view geographic metadata:

1. `view_geographic_paths()` - View all fully-specified geographic paths that are supported by a dataset of interest during the calendar year of interest.

<br>

2. `check_path_existence()` - Check if the specifier, or collection of specifiers, is supported for the dataset of interest during the specified calendar year. If supported, returns a list of fully-specified geographic paths containing at least one of the supplied specifiers anywhere.

```python
import pprint

import acspsuedo.query as apq
from acspsuedo.datasets import ACS1

YEAR = 2021
```

View all possible geographic paths

```python
all_paths = apq.view_geographic_paths(ACS1, YEAR)

pprint.pprint(all_paths)
```

View geographic paths containing a specifier (or specifiers)

```python
# Simulate a case where a user is interesteed in all paths
# for a given geographic scope (e.g. 'state' in our case)
paths = apq.check_path_existence(ACS1, YEAR, 'state')

pprint.pprint(paths)
```

```python
# Simulate a case when an unknown path is presented
apq.check_path_existence(ACS1, YEAR, ['asd', 'tract'])
```
